```python
def my_function(a, b):
    add = a + b
    sub = a - b
    mult = a * b
    div = a / b
    rem = a % b

    print("Addition: ", add)
    print("Subtraction: ", sub)
    print("Multiplication: ", mult)
    print("Division: ", div)
    print("Remainder: ", rem)

my_function(5, 2)
```

```python
from statistics import mean, median, mode, stdev
data = [1, 2, 3, 4, 5, 6, 7, 8, 9]

result = median(data)
print("Median is: ", result)
print("Mean is: ", mean(data))
print("Mode is: ", mode(data))
print("Standard Deviation is: ", stdev(data))
```
